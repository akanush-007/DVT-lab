```
# Correlation (numeric columns)
numeric_df = df.select_dtypes(include='number')
if not numeric_df.empty:
    numeric_df.corr()
else:
    print("No numeric columns found")
```

```
# Histogram of numeric columns
numeric_df = df.select_dtypes(include='number')
if not numeric_df.empty:
    numeric_df.hist(figsize=(12,8))
    plt.tight_layout()
    plt.show()
else:
    print("No numeric columns found")
```
