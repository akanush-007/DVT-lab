```python
import pandas as pd
df = pd.read_csv("fifa_world_cup_2026_player_performance.csv")
print("Missing values present per record/column:")
print(df.isnull())
```

```python

missing_counts = df.isnull().sum()
print("Missing Values Count:")
print(missing_counts[missing_counts > 0] if missing_counts.sum() > 0 else "No missing values found across columns.")
```

```python

df["player_rating"] = df["player_rating"].fillna(df["player_rating"].mean())
df["market_value_eur"] = df["market_value_eur"].fillna(df["market_value_eur"].mean())

print("Data after filling missing values:")
print(df[["player_name", "player_rating", "market_value_eur"]].head())
```

```python

print("Duplicate records found:", df.duplicated(subset=['player_id', 'match_id']).sum())

df.drop_duplicates(subset=['player_id', 'match_id'], inplace=True)
print("Total records after removing duplicates:", len(df))
```

```python

Q1 = df['market_value_eur'].quantile(0.25)
Q3 = df['market_value_eur'].quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[(df['market_value_eur'] < lower_limit) | (df['market_value_eur'] > upper_limit)]
print("Number of market value outliers detected:", len(outliers))
print("\nSample Outliers:")
print(outliers[['player_name', 'team', 'market_value_eur']].head())
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[['market_value_normalized']] = scaler.fit_transform(df[['market_value_eur']])
print("Normalized Market Values (0 to 1 scale):")
print(df[['player_name', 'market_value_eur', 'market_value_normalized']].head())
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df['position_encoded'] = encoder.fit_transform(df['position'])

print("Label Encoded Player Positions:")
print(df[['position', 'position_encoded']].drop_duplicates().reset_index(drop=True))
```

```python

preferred_foot_encoded = pd.get_dummies(df['preferred_foot'], prefix='foot')
print("One-Hot Encoded Preferred Foot:")
print(preferred_foot_encoded.head())
```
