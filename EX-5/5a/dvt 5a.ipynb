```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df=pd.read_csv("../../data/fifa_world_cup_2026_player_performance.csv"); print("Dataset loaded successfully"); print("Shape:",df.shape); display(df.head())
```

```python
player_goals=df.groupby("player_name")["goals"].sum().nlargest(10).sort_values(); display(player_goals)
```

```python
plt.figure(figsize=(9,5)); player_goals.plot(kind="barh"); plt.title("Top 10 Players – Goals"); plt.xlabel("Goals"); plt.ylabel("Player"); plt.tight_layout(); plt.show()
```

```python
country_goals=df.groupby("team")["goals"].sum().nlargest(15); plt.figure(figsize=(10,5)); plt.plot(country_goals.index,country_goals.values,marker="o"); plt.title("Top 15 Teams by Total Goals"); plt.xlabel("Team"); plt.ylabel("Total Goals"); plt.xticks(rotation=45,ha="right"); plt.grid(True,alpha=.3); plt.tight_layout(); plt.show()
```

```python
positions=df.groupby("position")["goals"].sum(); plt.figure(figsize=(7,7)); plt.pie(positions.values,labels=positions.index,autopct="%1.1f%%",startangle=90); plt.title("Goal Contribution by Position"); plt.tight_layout(); plt.show()
```

```python
plt.figure(figsize=(8,5)); plt.hist(df["goals"],bins=range(0,7),edgecolor="black",align="left"); plt.title("Distribution of Goals per Match Record"); plt.xlabel("Goals"); plt.ylabel("Number of Records"); plt.xticks(range(0,6)); plt.tight_layout(); plt.show()
```

```python
sample=df.sample(5000,random_state=42); plt.figure(figsize=(8,5)); plt.scatter(sample["shots"],sample["goals"],alpha=.25); plt.title("Shots vs Goals"); plt.xlabel("Shots"); plt.ylabel("Goals"); plt.tight_layout(); plt.show()
```
